```python
def ks_sample_size(alpha=0.05, power=0.8, effect_size=eff_size):
    """
    Estimates the required sample size for a two-sample Kolmogorov-Smirnov test.
    
    Parameters:
    alpha (float): Significance level (Type I error rate)
    power (float): Desired power (1 - Type II error rate)
    effect_size (float): Expected KS effect size (difference in distributions)
    
    Returns:
    int: Estimated required sample size per group.
    """
    # Compute critical values
    c_alpha = np.sqrt(-0.5 * np.log(alpha / 2))
    c_beta = np.sqrt(-0.5 * np.log(1 - power))
    
    # Compute required sample size per group
    n = ((c_alpha + c_beta) / effect_size) ** 2
    
    return int(np.ceil(n))

# Example usage
effect_size = 0.2  # Typical values: 0.1 (small), 0.2 (medium), 0.3+ (large)
n_required = ks_sample_size(alpha=0.05, power=0.8, effect_size=effect_size)
print(f"Estimated required sample size per group: {n_required}")
```
