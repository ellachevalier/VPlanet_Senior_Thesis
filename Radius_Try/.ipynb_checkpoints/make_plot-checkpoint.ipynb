```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dirs = []
for file in os.listdir("data_working/"):
    d = os.path.join("data_working/", file)
    if os.path.isdir(d):
        dirs.append(d)
        
sorted_dirs = sorted(dirs)
sorted_dirs

final_radii=[]
initial_radii=[]
good_radii=[]
#bad_indices=[]
plt.figure() 
count = 0
for path in sorted_dirs:
    if os.path.exists(path+'/system3.planet.forward'):
        forward_file = os.path.join(path, 'system3.planet.forward')
        df=pd.read_csv(forward_file, sep=' ', header=None)
        time = df[0]
        mass = df[1]
        env_mass = df[2]
        radius = df[3]
        final_radii.append(radius.tolist()[-1])
        initial_radii.append(radius.tolist()[0])
        if radius.tolist()[-1] != radius.tolist()[0]:
            good_radii.append(radius.tolist()[-1])
#         else:
#             bad_indices.append(indices[count])
        plt.plot(time, radius)
        count += 1
plt.ylabel("Planet radius (R_earth)")
plt.xlabel("Time (years)")
plt.title("Radius over time with atmospheric escape")
plt.show()

plt.figure()

plt.hist(good_radii, bins=50, color='skyblue', edgecolor='black') #orb period 20, planet mass loc=3, scale=10, size=size
#plt.xlim(0,4)
# Add titles and labels
plt.title('Distribution of Planet Radii')
plt.xlabel('Radius [R_earth]')
plt.ylabel('Frequency')

# Show the plot
plt.show()

`
```
